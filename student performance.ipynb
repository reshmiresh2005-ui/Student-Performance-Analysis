```python
total_students = len(df)

print(total_students)
```

```python
passed_students = (df["Pass"] == 1).sum()

print(passed_students)
```

```python
failed_students = (df["Pass"] == 0).sum()

print(failed_students)
```

```python
pass_percentage = (passed_students / total_students) * 100

print(pass_percentage)
```

```python
fail_percentage = (failed_students / total_students) * 100

print(fail_percentage)
```

```python
passed_study_hours = df[df["Pass"] == 1]["Study_Hours"].mean()

print(passed_study_hours)
```

```python
failed_study_hours = df[df["Pass"] == 0]["Study_Hours"].mean()

print(failed_study_hours)
```

```python
passed_attendance = df[df["Pass"] == 1]["Attendance"].mean()
failed_attendance = df[df["Pass"] == 0]["Attendance"].mean()

print("Passed students:", passed_attendance)
print("Failed students:", failed_attendance)

```

```python
passed_marks = df[df["Pass"] == 1]["Previous_Marks"].mean()
failed_marks = df[df["Pass"] == 0]["Previous_Marks"].mean()

print("Passed students:", passed_marks)
print("Failed students:", failed_marks)
```

```python
passed_backlogs = df[df["Pass"] == 1]["Backlogs"].mean()
failed_backlogs = df[df["Pass"] == 0]["Backlogs"].mean()

print("Passed students:", passed_backlogs)
print("Failed students:", failed_backlogs)
```

```python
pass_group = df.groupby("Pass")[
    [
        "Study_Hours",
        "Attendance",
        "Assignments",
        "Previous_Marks",
        "Backlogs"
    ]
].mean()

print(pass_group)
```

```python
internet_pass_percentage = (
    df.groupby("Internet_Access")["Pass"]
    .mean() * 100
)

print(internet_pass_percentage)

```

```python
pass_fail_counts = df["Pass"].value_counts().sort_index()

plt.figure(figsize=(7, 5))

plt.bar(
    ["Failed", "Passed"],
    [
        pass_fail_counts.get(0, 0),
        pass_fail_counts.get(1, 0)
    ]
)

plt.title("Number of Passed and Failed Students")
plt.xlabel("Result")
plt.ylabel("Number of Students")

plt.show()
```

```python
print("\n45. STUDY HOURS HISTOGRAM")

plt.figure(figsize=(7, 5))

plt.hist(df["Study_Hours"], bins=10)

plt.title("Distribution of Study Hours")
plt.xlabel("Study Hours")
plt.ylabel("Number of Students")

plt.show()
```

```python
print(df["Study_Hours"].describe())
```

```python
average_study = df.groupby("Pass")["Study_Hours"].mean()

plt.figure(figsize=(7, 5))

plt.bar(
    ["Failed", "Passed"],
    [
        average_study.get(0, 0),
        average_study.get(1, 0)
    ]
)

plt.title("Average Study Hours of Passed and Failed Students")
plt.xlabel("Result")
plt.ylabel("Average Study Hours")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    df["Study_Hours"],
    df["Previous_Marks"]
)

plt.title("Study Hours vs Previous Marks")
plt.xlabel("Study Hours")
plt.ylabel("Previous Marks")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.plot(df["Study_Hours"].reset_index(drop=True))

plt.title("Study Hours of Students")
plt.xlabel("Student Number")
plt.ylabel("Study Hours")

plt.show()
```
